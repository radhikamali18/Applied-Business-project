```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Branded_Electronics_Business_Analytics_Project_Dataset.csv")
data.head()
```

part 2 basic eda

```python
# First 5 records
print("FIRST 5 RECORDS")
display(data.head())

# Last 5 records
print("LAST 5 RECORDS")
display(data.tail())

# Dataset shape
print("DATASET SHAPE:", data.shape)

# Column names
print("\nCOLUMN NAMES:")
print(data.columns.tolist())

# Data types
print("\nDATA TYPES:")
print(data.dtypes)

# Missing values
print("\nMISSING VALUES:")
print(data.isnull().sum())

# Duplicate records
print("\nDUPLICATE RECORDS:")
print(data.duplicated().sum())

# Unique values
print("\nUNIQUE VALUES:")
print(data.nunique())

# Basic statistics
print("\nBASIC STATISTICS:")
display(data.describe())
```

part 3 data preprocessing

```python
# Convert Date into datetime format
data["Date"] = pd.to_datetime(data["Date"])

# Fill missing Rating with median
data["Rating"] = data["Rating"].fillna(data["Rating"].median())

# Fill missing Marketing Spend with median
data["Marketing_Spend"] = data["Marketing_Spend"].fillna(
    data["Marketing_Spend"].median()
)

# Remove duplicate records
data = data.drop_duplicates()

# Check invalid values
print("Invalid Quantity:", (data["Quantity"] <= 0).sum())
print("Invalid Revenue:", (data["Revenue"] <= 0).sum())

# Final missing values
print("\nMissing Values After Cleaning:")
print(data.isnull().sum())

print("\nFinal Dataset Shape:")
print(data.shape)
```

```python
# Bar Chart: Revenue by Brand

brand_revenue = data.groupby("Brand")["Revenue"].sum()

plt.figure(figsize=(8,5))
brand_revenue.plot(kind="bar")
plt.title("Revenue by Brand")
plt.xlabel("Brand")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

PART 7 — Correlation Analysis

```python
print("Dataset loaded successfully!")
print(data.shape)
variables = ["Price","Quantity","Revenue","Marketing_Spend","Rating","Demand"]

correlation = data[variables].corr()

print("CORRELATION MATRIX")
display(correlation)

plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

part 8 regression analysis

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Remove missing values
reg_data = data[["Marketing_Spend", "Revenue"]].dropna()

X = reg_data[["Marketing_Spend"]]
y = reg_data["Revenue"]

# Create model
model = LinearRegression()
model.fit(X, y)

# Prediction
y_pred = model.predict(X)

# Results
print("Regression Equation:")
print(f"Revenue = {model.intercept_:.2f} + ({model.coef_[0]:.2f} × Marketing Spend)")

print("\nCoefficient:", model.coef_[0])
print("R² Score:", r2_score(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y, y_pred)))

# Scatter + regression line
plt.figure(figsize=(8,5))
plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")
plt.show()
```
